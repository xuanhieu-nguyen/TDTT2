Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình

```python
import math

a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("Khong la ba canh cua mot tam giac")
else:
    # Phân loại
    if a == b == c:
        loai = "Tam giac deu"
    elif a == b or b == c or a == c:
        if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
            loai = "Tam giac vuong can"
        else:
            loai = "Tam giac can"
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        loai = "Tam giac vuong"
    else:
        loai = "Tam giac thuong"

    # Tính chu vi, diện tích
    chu_vi = a + b + c
    p = chu_vi / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))

    print("Loai tam giac:", loai)
    print("Chu vi:", chu_vi)
    print("Dien tich:", round(dien_tich, 2))
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
import math

a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))

if a == 0:
    # Quy về bậc nhất: bx + c = 0
    if b != 0:
        print("Phuong trinh bac nhat, nghiem: x =", -c / b)
    elif c != 0:
        print("Phuong trinh vo nghiem")
    else:
        print("Phuong trinh vo so nghiem")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        print("Phuong trinh co nghiem kep: x =", -b / (2*a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print("Phuong trinh co hai nghiem phan biet:")
        print("x1 =", x1)
        print("x2 =", x2)
```

Bài 8: Giải hệ phương trình tuyến tính:
ax + by = m
cx + dy = n

```python
a, b, m = map(float, input("Nhập a, b, m (ax + by = m): ").split())
c, d, n = map(float, input("Nhập c, d, n (cx + dy = n): ").split())

D = a*d - b*c
Dx = m*d - b*n
Dy = a*n - m*c

if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"He co nghiem duy nhat: x = {x:.2f}, y = {y:.2f}")
elif Dx != 0 or Dy != 0:
    print("He vo nghiem")
else:
    print("He vo so nghiem")
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây. 

```python
x = int(input("Nhập số giây: "))

gio = x // 3600
con_lai = x % 3600
phut = con_lai // 60
giay = con_lai % 60

print(f"{x} giây = {gio} giờ {phut} phút {giay} giây")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
x0 = float(input("Nhập hoành độ tâm x0: "))
y0 = float(input("Nhập tung độ tâm y0: "))
R = float(input("Nhập bán kính R: "))

x = float(input("Nhập hoành độ điểm A: "))
y = float(input("Nhập tung độ điểm A: "))

d2 = (x - x0)**2 + (y - y0)**2

if d2 == R**2:
    print("Diem A thuoc duong tron")
else:
    print("Diem A khong thuoc duong tron")
```
