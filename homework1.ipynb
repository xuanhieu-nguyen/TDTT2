```python
#Kiểm tra số chẵn lẻ: Viết chương trình nhập một số nguyên, kiểm tra và in ra số đó là số chẵn hay số lẻ.

n = int(input("Nhap vao so nguyen n :"))
if(n % 2 == 0):
 print("n la so nguyen chan")
else:
 print("n la so nguyen le")
```

```python
#Kiểm tra số có tận cùng 5 không: Viết chương trình nhập một số nguyên, kiểm tra số đó có tận cùng 5 không, nếu đúng in ra “True”, nếu sai in ra “False”.

n = int(input("Nhap vao so nguyen n :"))
if(abs(n) % 10 == 5):
 print("True")
else: print("False")
```

```python
#Kiểm tra tuổi đủ điều kiện bầu cử: Viết chương trình nhập năm sinh của một người, kiểm tra xem người đó có đủ 18 tuổi để đi bầu cử hay không. 

nam_hien_tai = int(input("Nhap nam hien tai : "))
nam_sinh = int(input("Nhap vao nam sinh cua cu tri :"))
tuoi = nam_hien_tai - nam_sinh
if(tuoi < 18):
    print("Khong du dieu kien bau cu")
else: 
    print("Du dieu kien bau cu")
```

```python
#So sánh 2 số: Viết chương trình nhập vào hai số, in ra số lớn hơn hoặc thông báo nếu hai số bằng nhau. 

a = int(input("Nhap vao so a :"))
b = int(input("Nhap vao so b :"))
if(a == b):
    print("Hai so bang nhau")
elif(a != b ):
    if(a > b):
        print(a)
    else:
        print(b)
```

```python
#Kiểm tra ký tự nhập vào là chữ hay số: Viết chương trình nhập một ký tự, kiểm tra xem ký tự đó là chữ cái hay số. 
kytu = input("Nhap vao mot ky tu: ")

if kytu.isdigit():
    print("Day la chu so.")
elif kytu.isalpha():
    print("Day la chu cai.")
else:
    print("Day la ky tu dac biet.")
```

```python
#Tính điểm học lực Nhập điểm (0-10), phân loại học lực theo quy tắc: >=8: Giỏi; >=6.5 và <8: Khá; >=5 và <6.5: Trung bình; <5: Yếu 
diem_so = float(input("Nhap diem so cua hoc sinh :"))
if(diem_so >= 8):
    print("Gioi")
elif(diem_so >= 6.5 and diem_so < 8):
    print("Kha")
elif(diem_so >= 5 and diem_so < 6.5):
    print("Trung binh")
else:
    print("Yeu")
 
```

```python
#Kiểm tra năm nhuận: Viết chương trình nhập năm, kiểm tra xem năm đó có phải năm nhuận hay không. (Năm nhuận nếu chia hết cho 400 hoặc chia hết cho 4 nhưng không chia hết cho 100) 
nhap_nam = int(input("Nhap mot nam bat ky"))
if(nhap_nam % 4 == 0 and nhap_nam % 100 != 0):
    print("Nam nhuan")
else:
    print("Nam khong nhuan")
```

```python
#Chuyển đổi số thành chữ số: Nhập một số nguyên từ 0 đến 9, in ra tên chữ số tương ứng (0 -> "Không", 1 -> "Một", ...) – Xem trước match …case
num = int(input("Nhap vao mot so nguyen duong tu 0 den 9 :"))
match num:
    case 0:
        print("Không")
    case 1:
        print("Một")
    case 2:
        print("Hai")
    case 3:
        print("Ba")
    case 4:
        print("Bốn")
    case 5:
        print("Năm")
    case 6:
        print("Sáu")
    case 7:
        print("Bảy")
    case 8:
        print("Tám")
    case _:
        print("Chín")
 
```
